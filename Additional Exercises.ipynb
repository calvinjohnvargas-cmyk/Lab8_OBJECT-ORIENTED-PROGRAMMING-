```python
class Book:
    def __init__(self, title, author, is_available=True):
        self.title = str(title).strip()
        self.author = str(author).strip()
        self.is_available = is_available

    def __str__(self):
        status = "Available" if self.is_available else "Borrowed"
        return f"'{self.title}' by {self.author} - Status: {status}"

class Library:
    def __init__(self):
        self.books = {}

    def add_book(self, title, author, is_available=True):
        title = str(title).strip()
        if title in self.books:
            print(f"-> Book '{title}' already exists in the library.")
            return
        self.books[title] = Book(title, author, is_available)
        print(f"-> Successfully added '{title}'.")

    def borrow_book(self, title):
        title = str(title).strip()
        if title not in self.books:
            print(f"-> Error: The book '{title}' does not exist in our system.")
            return

        book = self.books[title]
        if not book.is_available:
            print(f"-> Error: The book '{title}' is already checked out.")
        else:
            book.is_available = False
            print(f"-> Successfully borrowed '{title}'.")

    def return_book(self, title):
        title = str(title).strip()
        if title not in self.books:
            print(f"-> Error: The book '{title}' does not belong to this library.")
            return

        book = self.books[title]
        if book.is_available:
            print(f"-> Notice: The book '{title}' was already available.")
        else:
            book.is_available = True
            print(f"-> Successfully returned '{title}'.")

    def display_books(self):
        if not self.books:
            print("\n--- The library is currently empty. ---")
            return
        
        print("\n--- Library Collection ---")
        for book in self.books.values():
            print(book)
        print("--------------------------")

def main():
    lib = Library()

    while True:
        try:
            num_books = int(input("Enter the number of books to add initially: "))
            if num_books < 0:
                print("Please enter a positive number.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a whole number.")

    # Loop to add the initial books
    for i in range(num_books):
        print(f"\nAdding Book {i+1}:")
        title = input("Enter title: ")
        author = input("Enter author: ")
        
        # Determine availability status
        while True:
            status_input = input("Is it currently available? (y/n): ").strip().lower()
            if status_input in ['y', 'yes']:
                is_available = True
                break
            elif status_input in ['n', 'no']:
                is_available = False
                break
            else:
                print("Invalid input. Please enter 'y' or 'n'.")
        
        lib.add_book(title, author, is_available)

    # Interactive Menu for ongoing management
    while True:
        print("\n=== Library Menu ===")
        print("1. Display all books")
        print("2. Borrow a book")
        print("3. Return a book")
        print("4. Add a new book")
        print("5. Exit")
        
        choice = input("Enter your choice (1-5): ").strip()
        
        if choice == '1':
            lib.display_books()
        elif choice == '2':
            title = input("Enter the title of the book to borrow: ")
            lib.borrow_book(title)
        elif choice == '3':
            title = input("Enter the title of the book to return: ")
            lib.return_book(title)
        elif choice == '4':
            print("\nAdding a new book:")
            title = input("Enter title: ")
            author = input("Enter author: ")
            lib.add_book(title, author)
        elif choice == '5':
            print("Exiting Library System. Goodbye!")
            break
        else:
            print("Invalid choice. Please select a number from 1 to 5.")

if __name__ == "__main__":
    main()
```

```python
def get_positive_number(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Invalid input. Value must not be negative.")
            else:
                return value
        except ValueError:
            print("Invalid input. Enter a number.")


def main():
    while True:
        try:
            num_employees = int(input("Enter number of employees: "))
            if num_employees <= 0:
                print("Enter at least one employee.")
            else:
                break
        except ValueError:
            print("Invalid input. Enter a whole number.")

    employees = []

    for i in range(num_employees):
        print(f"\nEmployee {i+1}")
        name = input("Name: ")

        hourly_rate = get_positive_number("Hourly rate: ")
        hours_worked = get_positive_number("Hours worked: ")

        overtime_hours = 0
        overtime_pay = 0

        if hours_worked > 160:
            overtime_hours = hours_worked - 160
            overtime_pay = overtime_hours * hourly_rate * 1.5
            base_hours = 160
        else:
            base_hours = hours_worked

        base_salary = base_hours * hourly_rate

        employees.append({
            "name": name,
            "hours": hours_worked,
            "base_salary": base_salary,
            "overtime_pay": overtime_pay
        })

    print("\nPayroll Summary")
    print("-" * 40)

    for emp in employees:
        print(f"Name: {emp['name']}")
        print(f"Hours Worked: {emp['hours']}")
        print(f"Base Salary: {emp['base_salary']:.2f}")
        print(f"Overtime Pay: {emp['overtime_pay']:.2f}")
        print("-" * 40)


if __name__ == "__main__":
    main()
```

```python
products = {
    1: {"name": "Laptop", "price": 50000, "stock": 5},
    2: {"name": "Mouse", "price": 500, "stock": 20},
    3: {"name": "Keyboard", "price": 1500, "stock": 10},
    4: {"name": "Headphones", "price": 2000, "stock": 8}
}

cart = {}

def display_products():
    print("\nAvailable Products")
    print("-" * 40)
    for pid, item in products.items():
        print(f"{pid}. {item['name']} | Price: {item['price']} | Stock: {item['stock']}")


def display_cart():
    print("\nShopping Cart")
    print("-" * 40)
    total = 0

    if not cart:
        print("Cart is empty")
        return

    for pid, qty in cart.items():
        name = products[pid]["name"]
        price = products[pid]["price"]
        subtotal = price * qty
        total += subtotal

        print(f"{name} | Qty: {qty} | Subtotal: {subtotal}")

    print("-" * 40)
    print(f"Total Price: {total}")


def add_to_cart():
    display_products()
    try:
        pid = int(input("Enter product number to add: "))
        if pid not in products:
            print("Invalid product.")
            return

        qty = int(input("Enter quantity: "))
        if qty <= 0:
            print("Invalid quantity.")
            return

        if qty > products[pid]["stock"]:
            print("Not enough stock.")
            return

        products[pid]["stock"] -= qty
        cart[pid] = cart.get(pid, 0) + qty

        print("Item added to cart.")

    except ValueError:
        print("Invalid input.")


def remove_from_cart():
    display_cart()
    try:
        pid = int(input("Enter product number to remove: "))
        if pid not in cart:
            print("Item not in cart.")
            return

        qty = int(input("Enter quantity to remove: "))
        if qty <= 0 or qty > cart[pid]:
            print("Invalid quantity.")
            return

        cart[pid] -= qty
        products[pid]["stock"] += qty

        if cart[pid] == 0:
            del cart[pid]

        print("Item removed from cart.")

    except ValueError:
        print("Invalid input.")


def main():
    while True:
        print("\n1. Add Item")
        print("2. Remove Item")
        print("3. View Cart")
        print("4. Exit")

        choice = input("Choose option: ")

        if choice == "1":
            add_to_cart()
        elif choice == "2":
            remove_from_cart()
        elif choice == "3":
            display_cart()
        elif choice == "4":
            print("Checkout complete.")
            break
        else:
            print("Invalid choice.")


if __name__ == "__main__":
    main()
```
