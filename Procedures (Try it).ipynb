```python
class Laptop:
    def __init__(self, brand, processor, ram):
        self.brand = brand
        self.processor = processor
        self.ram = ram

    def get_specs(self):
        return f"Laptop Specs: [Brand: {self.brand}, Processor: {self.processor}, RAM: {self.ram}GB]"

my_laptop = Laptop("Apple", "M3 Max", 32)

print(my_laptop.get_specs())
```

```python
class StudentRecord:
    def __init__(self):
        self.__grades = [] 

    def add_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grades.append(grade)
        else:
            print("Invalid grade! Please enter a value between 0 and 100.")

    def get_all_grades(self):
        return self.__grades

my_record = StudentRecord()

my_record.add_grade(85)
my_record.add_grade(92)
my_record.add_grade(78)

print(f"Student Grades: {my_record.get_all_grades()}")
```

```python
class Device:
    def __init__(self, device_name):
        self.device_name = device_name

    def power_on(self):
        return f"The {self.device_name} is now powered on."

class Smartphone(Device):
    def power_on(self):
        return f"The {self.device_name} is booting... Welcome to your Smart Life!"

generic_device = Device("Office Monitor")
my_phone = Smartphone("iPhone 15")

print(generic_device.power_on())
print(my_phone.power_on())
```

```python
class Device:
    def power_on(self):
        return "Generic device powering on..."

class Smartphone(Device):
    def power_on(self):
        return "Smartphone: Screen glowing, vibration active. Welcome!"

class Laptop(Device):
    def power_on(self):
        return "Laptop: Cooling fans spinning, system booting up..."

def start_device(device):
    print(device.power_on())

my_phone = Smartphone()
my_laptop = Laptop()

start_device(my_phone)
start_device(my_laptop)
```

```python
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9 / 5) + 32

user_input = float(input("Enter a temperature in Celsius: "))
fahrenheit = TemperatureConverter.celsius_to_fahrenheit(user_input)

print(f"{user_input}°C is equal to {fahrenheit}°F")
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(10, 5)
v2 = Vector(4, 3)

addition_result = v1 + v2
subtraction_result = v1 - v2

print(f"Vector 1: {v1}")
print(f"Vector 2: {v2}")
print(f"Addition (v1 + v2): {addition_result}")
print(f"Subtraction (v1 - v2): {subtraction_result}")
```

```python
from abc import ABC, abstractmethod

class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

class Refrigerator(Appliance):
    def turn_on(self):
        return "Refrigerator is now cooling."

my_fridge = Refrigerator()
print(my_fridge.turn_on())
```

```python
class A:
    def show(self):
        print("Identifying: Class A")

class B(A):
    def show(self):
        print("Identifying: Class B")

class C(A):
    def show(self):
        print("Identifying: Class C")

class D(B, C):
    def show(self):
        print("Identifying: Class D")

print("--- Method Resolution Order for D ---")

for cls in D.mro():
    print(cls)

obj = D()
print("\n--- Calling obj.show() ---")
obj.show()
```
